```python
import pandas as pd
import numpy as np

df = pd.read_csv("train.csv", parse_dates=["date"])

# Rank stores and items by average sales
store_rank = df.groupby("store")["sales"].mean().sort_values().index.tolist()
item_rank = df.groupby("item")["sales"].mean().sort_values().index.tolist()

# 5 stores spread from low to high, 10 items spread from low to high
stores = [store_rank[i] for i in [0, 2, 4, 6, 9]]
items = [item_rank[i] for i in np.linspace(0, 49, 10).astype(int)]

print("Selected stores:", stores)
print("Selected items:", items)

subset = df[df["store"].isin(stores) & df["item"].isin(items)]
print(subset.shape)   # expect (5 * 10 * 1826 = 91300, 4)

subset.to_csv("data_subset.csv", index=False)
```
